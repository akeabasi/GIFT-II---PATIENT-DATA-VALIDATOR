```python

# TASK 1: PATIENT VARIABLES


patient_name = "David Okoro"
patient_id = "PT20260801"
age = 45
gender = "Male"
temperature = 37.2
heart_rate = 82
emergency_status = False



# TASK 2, 3, 4, 5 AND 6:
# INDIVIDUAL VALIDATION FUNCTIONS


def validate_name(name):
    if name.strip() != "":
        return True
    else:
        return False


def validate_patient_id(patient_id):
    if patient_id.strip() != "":
        return True
    else:
        return False


def validate_age(age):
    if age > 0 and age <= 120:
        return True
    else:
        return False


def validate_gender(gender):
    if gender == "Male":
        return True
    elif gender == "Female":
        return True
    elif gender == "Other":
        return True
    else:
        return False


def validate_temperature(temperature):
    if temperature >= 35 and temperature <= 42:
        return True
    else:
        return False


def validate_heart_rate(heart_rate):
    if heart_rate > 40 and heart_rate < 200:
        return True
    else:
        return False



# TASK 7: AGE CATEGORY FUNCTION


def get_age_category(age):

    if age >= 0 and age <= 12:
        return "Child"

    elif age >= 13 and age <= 17:
        return "Teenager"

    elif age >= 18 and age <= 64:
        return "Adult"

    elif age >= 65:
        return "Senior"

    else:
        return "Invalid Age"



# TASK 9: OVERALL VALIDATION FUNCTION


def validate_patient(name, patient_id, age, gender, temperature, heart_rate):

    name_valid = validate_name(name)
    patient_id_valid = validate_patient_id(patient_id)
    age_valid = validate_age(age)
    gender_valid = validate_gender(gender)
    temperature_valid = validate_temperature(temperature)
    heart_rate_valid = validate_heart_rate(heart_rate)

    if (name_valid and
            patient_id_valid and
            age_valid and
            gender_valid and
            temperature_valid and
            heart_rate_valid):

        return True

    else:
        return False


# ----------------------------------------
# RUN INDIVIDUAL VALIDATIONS
# ----------------------------------------

name_valid = validate_name(patient_name)

patient_id_valid = validate_patient_id(patient_id)

age_valid = validate_age(age)

gender_valid = validate_gender(gender)

temperature_valid = validate_temperature(temperature)

heart_rate_valid = validate_heart_rate(heart_rate)


# ----------------------------------------
# TASK 7: DETERMINE AGE CATEGORY
# ----------------------------------------

age_category = get_age_category(age)


# ----------------------------------------
# TASK 8: OVERALL VALIDATION STATUS
# ----------------------------------------

record_valid = validate_patient(
    patient_name,
    patient_id,
    age,
    gender,
    temperature,
    heart_rate)


# ----------------------------------------
# TASK 11: FOR LOOP FOR VALIDATION RESULTS
# ----------------------------------------

validation_results = [
    name_valid,
    patient_id_valid,
    age_valid,
    gender_valid,
    temperature_valid,
    heart_rate_valid]

validation_names = [ "Name Valid", "Patient ID Valid", "Age Valid", "Gender Valid", "Temperature Valid", "Heart Rate Valid"]


print("========================================")
print("       PATIENT DATA VALIDATION REPORT")
print("========================================")


print("Patient Name:", patient_name)
print("Patient ID:", patient_id)
print("Age:", age)
print("Gender:", gender)
print("Temperature:", temperature, "°C")
print("Heart Rate:", heart_rate, "BPM")
print("Emergency Status:", emergency_status)


print("----------------------------------------")
print("VALIDATION RESULTS")
print("----------------------------------------")



# Using range() with the for loop
for index in range(len(validation_results)):
    print(validation_names[index] + ":", validation_results[index])



print("Age Category:", age_category)


print("----------------------------------------")
print("FINAL STATUS")
print("----------------------------------------")


if record_valid:
    print("Patient Record: Valid")
else:
    print("Patient Record: Invalid")


print("========================================")
```

```python
## ========================================
# BONUS: MULTIPLE PATIENT VALIDATION


def validate_name(name):
    if name.strip() != "":
        return True
    else:
        return False


def validate_patient_id(patient_id):
    if patient_id.strip() != "":
        return True
    else:
        return False


def validate_age(age):
    if age > 0 and age <= 120:
        return True
    else:
        return False


def validate_gender(gender):
    if gender == "Male":
        return True

    elif gender == "Female":
        return True

    elif gender == "Other":
        return True

    else:
        return False


def validate_temperature(temperature):
    if temperature >= 35 and temperature <= 42:
        return True
    else:
        return False


def validate_heart_rate(heart_rate):
    if heart_rate > 40 and heart_rate < 200:
        return True
    else:
        return False


def get_age_category(age):

    if age >= 0 and age <= 12:
        return "Child"

    elif age >= 13 and age <= 17:
        return "Teenager"

    elif age >= 18 and age <= 64:
        return "Adult"

    elif age >= 65 and age <= 120:
        return "Senior"

    else:
        return "Invalid Age"


# ----------------------------------------
# PATIENT RECORDS
# ----------------------------------------

patient_names = ["David Okoro", "Grace Bello", "Samuel Ade", "Anita James"]

patient_ids = ["PT001", "PT002", "", "PT004"]

patient_ages = [45, 72, 15, 130]

patient_genders = ["Male", "Female", "Male", "Female"]

patient_temperatures = [37.2, 36.8, 38.5, 36.9]

patient_heart_rates = [82, 78, 105, 76]


# ----------------------------------------
# COUNTERS
# ----------------------------------------

valid_records = 0
invalid_records = 0

invalid_name_count = 0
invalid_id_count = 0
invalid_age_count = 0
invalid_gender_count = 0
invalid_temperature_count = 0
invalid_heart_rate_count = 0


# ----------------------------------------
# PROCESS PATIENTS
# ----------------------------------------

for index in range(len(patient_names)):

    name = patient_names[index]
    patient_id = patient_ids[index]
    age = patient_ages[index]
    gender = patient_genders[index]
    temperature = patient_temperatures[index]
    heart_rate = patient_heart_rates[index]

    print("========================================")
    print("PATIENT", index + 1)
    print("========================================")

    print("Name:", name)
    print("Patient ID:", patient_id)
    print("Age:", age)
    print("Gender:", gender)
    print("Temperature:", temperature, "°C")
    print("Heart Rate:", heart_rate, "BPM")

    print("VALIDATION RESULTS")

    name_valid = validate_name(name)
    patient_id_valid = validate_patient_id(patient_id)
    age_valid = validate_age(age)
    gender_valid = validate_gender(gender)
    temperature_valid = validate_temperature(temperature)
    heart_rate_valid = validate_heart_rate(heart_rate)

    print("Name Valid:", name_valid)
    print("Patient ID Valid:", patient_id_valid)
    print("Age Valid:", age_valid)
    print("Gender Valid:", gender_valid)
    print("Temperature Valid:", temperature_valid)
    print("Heart Rate Valid:", heart_rate_valid)

    print("Age Category:", get_age_category(age))

    
    # Count failed validation rules

    if not name_valid:
        invalid_name_count += 1

    if not patient_id_valid:
        invalid_id_count += 1

    if not age_valid:
        invalid_age_count += 1

    if not gender_valid:
        invalid_gender_count += 1

    if not temperature_valid:
        invalid_temperature_count += 1

    if not heart_rate_valid:
        invalid_heart_rate_count += 1


    # Determine overall record status

    if (name_valid and
            patient_id_valid and
            age_valid and
            gender_valid and
            temperature_valid and
            heart_rate_valid):

        print("Patient Record: Valid")
        valid_records += 1

    else:
        print("Patient Record: Invalid")
        invalid_records += 1


# ----------------------------------------
# FINAL SUMMARY
# ----------------------------------------

print("========================================")
print("FINAL VALIDATION SUMMARY")
print("========================================")

print("Valid Patient Records:", valid_records)
print("Invalid Patient Records:", invalid_records)

print("VALIDATION RULE FAILURES")

print("Invalid Names:", invalid_name_count)
print("Invalid Patient IDs:", invalid_id_count)
print("Invalid Ages:", invalid_age_count)
print("Invalid Genders:", invalid_gender_count)
print("Invalid Temperatures:", invalid_temperature_count)
print("Invalid Heart Rates:", invalid_heart_rate_count)

print("========================================")
```
